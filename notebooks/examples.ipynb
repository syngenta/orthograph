```python
import json

from orthograph.graph_data_model import GraphDataModel, ValidationConfig
from orthograph.graph_schema import GraphSchema, NodeSpec, RelationshipSpec
```

# Social Network

```python
# Define the schema
schema = GraphSchema(
    name="SocialNetwork",
    node_specs={
        "Person": NodeSpec(
            label="Person",
            properties={"name": "str", "age": "int"},
            uid_field="name",
            uid_type="str",
        ),
        "Movie": NodeSpec(
            label="Movie",
            properties={"title": "str", "year": "int"},
            uid_field="title",
            uid_type="str",
        ),
    },
    relationship_specs={
        "ACTED_IN": RelationshipSpec(
            label="ACTED_IN",
            node_types=("Person", "Movie"),
            properties={"role": "str"},
        ),
        "DIRECTED": RelationshipSpec(label="DIRECTED", node_types=("Person", "Movie")),
    },
)

# Create a GraphDataModel instance
config = ValidationConfig(
    allow_unknown_node_labels=False,
    allow_unknown_relationship_types=False,
    allow_extra_node_properties=True,
    allow_extra_relationship_properties=False,
)
data_model = GraphDataModel(schema, config)

# Test node validation
print("Validating nodes:")
node_data_list = [
    {"label": "Person", "name": "Tom Hanks", "age": 64},
    {"label": "Movie", "title": "Forrest Gump", "year": 1994},
    {
        "label": "Person",
        "name": "Emma Watson",
        "age": 30,
        "country": "UK",
    },  # Extra property
    {"label": "City", "name": "New York"},  # Unknown label
    {"label": "Person", "age": 40},  # Missing required property
]

for node_data in node_data_list:
    validated_node, errors = data_model.validate_node(node_data, collect_errors=True)
    if validated_node:
        print(f"Valid node: {validated_node}")
    if errors:
        print(f"Validation errors: {errors}")
    print()

# Test relationship validation
print("\nValidating relationships:")
rel_data_list = [
    {
        "type": "ACTED_IN",
        "start_node": "Tom Hanks",
        "end_node": "Forrest Gump",
        "role": "Forrest",
    },
    {
        "type": "DIRECTED",
        "start_node": "Robert Zemeckis",
        "end_node": "Forrest Gump",
    },
    {
        "type": "ACTED_IN",
        "start_node": "Emma Watson",
        "end_node": "Harry Potter",
        "role": "Hermione",
        "years": "2001-2011",
    },  # Extra property
    {
        "type": "FRIENDS_WITH",
        "start_node": "Tom Hanks",
        "end_node": "Emma Watson",
    },  # Unknown type
    {
        "type": "ACTED_IN",
        "start_node": "Tom Hanks",
        "end_node": "Forrest Gump",
    },  # Missing required property
]

for rel_data in rel_data_list:
    validated_rel, errors = data_model.validate_relationship(
        rel_data, collect_errors=True
    )
    if validated_rel:
        print(f"Valid relationship: {validated_rel}")
    if errors:
        print(f"Validation errors: {errors}")
    print()
```
